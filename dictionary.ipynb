Add, Remove & Change items

```python
my_dict1 = {'Name':'GS','id':12345,'DOB':1992,'Address':'HYD'}
my_dict1
```

```python
my_dict1['DOB']=2004 #Changing items 
my_dict1['Address']='New York'
my_dict1
```

```python
dict22 = {'DOB':2006} #Updating items
my_dict1.update(dict22)
my_dict1
```

```python
my_dict1.pop('id') #Removing keys
my_dict1
```

```python
my_dict1.popitem() # A random item is removed
```

```python
del[my_dict1['DOB']] 
my_dict1
```
