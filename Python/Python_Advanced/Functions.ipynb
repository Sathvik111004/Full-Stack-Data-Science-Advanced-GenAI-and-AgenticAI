```python
# Actual Argument divided into 4 types
# 1. Postion 
# 2. Keyword
# 3. Default
# 4. Variable length
```

```python
# 1 - Positional Argument

def person(name,age):
    print(name)
    print(age)
person('nit')

```

```python
def person(age,name):
    print(name)
    print(age)
person('nit',23) # here age = nit , name =1
```

```python

def person(age,name):
    print(name)
    print(age+1)
person('nit',23) # here age = nit , name =1
```

```python
# 2 - Keyword Argument

def person(age,name):
    print(name)
    print(age+1)
person(name='nit',age=23)
```

```python

def person(age,name,phno):
    print(name)
    print(age+1)
    print(phno)
person(name='nit',age=23,phno=12345)
```

```python
# 3 - Default Argument

def person(name,age=18): # Set default value for age
    print(name)
    print(age+1)
person('nit')
    
```

```python
# 4 - Variable Length Argument
def sum(a,*b):
    c=a+b
    print(c)
sum(5,6,7,8) # a = 5 , *b = 678
```

```python
def sum(a,*b):
    print(type(a))
    print(type(b))
    print((a))
    print((b))
sum(5,6,7,8)
```

```python
def sum(a,*b):
    c=a
    for i in b: # for loop for adding *b values
        c=c+i
    print(c)
sum(5,6,7,8,9) # 5+(6,7,8,9)
```

# KWARG ( KW + ARGS)



```python
def person(name,*data):
    print('name')
    print(data)
person('ALEX',age=36,home='south',mob=995599)
```

```python
def person(name,**data):
    print('name')
    print(data)
person('ALEX',age=36,home='south',mob=995599)
```

Single * --- ARGS
TWO ** --- KWARGS
