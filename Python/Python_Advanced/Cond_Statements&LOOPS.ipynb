LOOPS

While Loop

```python
i = 6        # Initialization
while i>=1:  # Condition
    print('Data Science',i)
    i = i - 1  #DECREMENT
```

```python
# Nested while
i =1
while i<=5:
    print('Data Science',end=' ') # Use end to print in same line
    j=1
    while j<=4:
        print('Tech',end=', ')
        j=j+1
    i = i+1
    print()
```

```python
i =1
while i<=4:
    j=0
    while j<=3:
        print(i*j, end=' ')
        j+=1
    print()
    i+=1
```

```python
# Range
for i in range(2,6):
    print(i)
```

```python
# Program to print 5 multiplication table
for i in range(1,51):
    if(i%5==0):
        print(i,end=' ')
```

```python
# I dont want 11 number , I want only 5 number from range 1 to 10 using break

for i in range(1,10):
    if i == 6:
        break
    print(i)
```

```python
# Continue 
for i in range(1,10):
    if i == 6:
        continue # Continue the loop except 6....
    print(i,end=' ')
```

```python
# Pass
for i in range(1,11):
```

```python
for i in range(1,11):
    pass # Ignores the path by using pass
```

```python
# Numbers from 1 to 50 and do not print the numbers divisible by 3 and 5
for i in range(1,51):
    if(i%3==0 or i%5==0):
        continue  
    print(i,end=' ')
    
```

For else

```python
nums = [12,13,4,77,99]
for num in nums:
    if num%5==0:
        print(num)
        break
else:
    print('Number not found')
```

```python
# Prime number

num = 13

for i in range(2, num):
    if num % i == 0:
        print("Not prime")
        break
else:
    print("Prime")
```
