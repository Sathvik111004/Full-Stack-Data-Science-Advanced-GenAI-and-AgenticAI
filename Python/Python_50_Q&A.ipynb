Q16 :  Write a program that checks if the input number is a prime number.

```python
 # Q16 ans
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("It is not a prime number")
            break
    else:
        print("It is a prime number")
else:
    print("It is not a prime number")
```

Q35:  How do you validate if an entered input is a valid email address?

```python
import re  # Q35 Ans
email = input("Enter email: ")
if re.match(r"[^@]+@[^@]+\.[^@]+", email):
    print("Valid email")
else:
    print("Invalid email")
```

Q48 : Write a program to count the number of consonants in a string entered by the

```python
c = input("Enter String: ") # Q48 ans

count = 0

for i in c:
    if i not in "aeiouAEIOU":
        count += 1

print("Number of consonants:", count)
```

Q49 : How do you check if a string entered by the user contains any punctuation?

```python
pun = input("Enter String: ") # Q49 ans

count = 0

for i in pun:
    if i  in ".,?;!'()[]-/@*":
        count += 1

print("Number of Punctuations?:", count)
```
