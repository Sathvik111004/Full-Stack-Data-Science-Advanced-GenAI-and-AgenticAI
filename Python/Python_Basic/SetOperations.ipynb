```python
# SuperSet , SubSet , Disjoint
```

```python
a2 ={1,2,3,4,5,6,7,8,9}
b2={3,4,5,6,7,8}
c2={10,20,30,40}
```

```python
a2.issuperset(b2) # a2 is dad, b2 has values of dad
```

```python
a2.isdisjoint(b2) # a2 and b2 are not neighbours
```

```python
b2.issubset(a2) # b2 contains values of a2
```

```python
b2.isdisjoint(c2) # Both contains diff values hence, neighbours 
```
