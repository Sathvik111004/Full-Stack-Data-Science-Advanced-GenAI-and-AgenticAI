```python
import numpy as np
```

```python
#Stack arrays vertically

a2 = np.array([1,2])
b2 = np.array([3,4])
stacked = np.vstack([a2,b2])
print(stacked)
```

```python
#Stack arrays horizontally
stacked = np.hstack([a2,b2])
print(stacked)
```

Mathematical Functions

```python
#Add 2 arrays
g = np.array([1,2,3,4])
added = np.add(g,2)
print(added)
```

```python
#Squared each element
square = np.power(g,2)
print(square)
```

```python
#Square root of each element
root = np.sqrt(g)
print(root)
```

```python
#Dot product of 2 arrays
a3=np.array([4,5,6,7])
dot = np.dot(g,a3)
print(dot)
```

Statistics Function

```python
#Mean of array
mean = np.mean(a3)
print(mean)
```

```python
# Variance of array
vari = np.var(a3)
print(a3)
```

```python
# Standard Deviation of Array
sd = np.std(a3)
print(sd)
```

```python
# Min element of array
mini = np.min(a3)
print(mini)
```

```python
# Max element of Array
maxi = np.max(a3)
print(maxi)
```

Random Sampling Function

```python
rand = np.random.rand(3) #Prints array of 3 random values
print(rand)
```

```python
np.random.seed(0) # Does not change the random values
randints = np.random.randint(0,10,size=6)
print(randints)
```

Boolean And Logical Functions

```python
# Check if all elements are true
#all
logi = np.array([True,False,True])
all_True = np.all(logi) # Check if all are true
print(all_True)
```

```python
# Check if all elements are true
#all
logi = np.array([False,False,False])
all_True = np.all(logi) # Check if all are true
print(all_True)
```

```python
#any
any_true = np.any(logi) # Check if any is True
print(any_true)
```

Set operations

```python
import numpy as np
set1 = np.array([1,2,3,4])
set2 = np.array([3,4,5,6])
inter = np.intersect1d(set1,set2)
print(inter)
```

```python
union = np.union1d(set1,set2)
print(union)
```

Array Attribute Functions

```python
# Array attributes
a = np.array([1, 2, 3])
shape = a.shape  # Shape of the array
size = a.size    # Number of elements
dimensions = a.ndim  # Number of dimensions
dtype = a.dtype   # Data type of the array

print("Shape of a:", shape)
print("Size of a:", size)
print("Number of dimensions of a:", dimensions)
print("Data type of a:", dtype)
```

Other Functions

```python
# Create a copy of an array
a = np.array([1, 2, 3])
copied_array = np.copy(a)  # Create a copy of array a
print("Copied array:", copied_array)
```

```python
# Size in bytes of an array
array_size_in_bytes = a.nbytes  # Size in bytes
print("Size of a in bytes:", array_size_in_bytes)
```

```python
# Check if two arrays share memory
shared = np.shares_memory(a, copied_array)  # Check if arrays share memory
print("Do a and copied_array share memory?", shared)
```
